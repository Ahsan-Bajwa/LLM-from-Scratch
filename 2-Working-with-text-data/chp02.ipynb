## 2.1 Reading a short story

```python
# Reading text from file and displaying it onto the screen 
with open("the-verdict.txt", 'r', encoding="utf-8") as f:
    raw_text = f.read()

print(f"Total number of characters: {len(raw_text)}")
print(f"{raw_text[:99]}")
```

```python
# Splitting text manually using re(Regular Expression) built-in python library
import re
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text) 
print(preprocessed[:99])
```

```python
# removing whitespaces to cut computing requirements
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(len(preprocessed))
print(preprocessed[:99])
```

```python
# sorting tokens in alphabetical order
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)
print(all_words[:30])
```

```python
# Creating vocab
vocab = {token:integer for integer,token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i>=50:
        break
```

```python
# Testing encoder and decoder part using sample text from the-verdict.txt
from SimpleTokenizerV1 import SimpleTokenizerV1

tokenizer = SimpleTokenizerV1(vocab)
text = "My dear, since I've chucked painting people don't say that stuff about me--they say it about Victor Grindle,"
encoded_text = tokenizer.encoder(text)
print(encoded_text)

decoded_text = tokenizer.decoder(encoded_text)
print(decoded_text)

# tokenizer = SimpleTokenizerV1(vocab)
# text = "Hello My dear, since I've chucked painting people don't say that stuff about me--they say it about Victor Grindle," ## This will produce Keyerror because there is no hello in the training text, vocab.
# encoded_text = tokenizer.encoder(text)
# print(encoded_text)

```

```python
tokens = sorted(list(set(preprocessed)))
tokens.extend(['<|unk|>','<|endoftext|>'])
vocab = {token:integer for integer, token in enumerate(tokens)}
print(len(vocab.items()))

for i, s in enumerate(list(vocab.items())[-5:]):
    print(s)
```

```python
from SimpleTokenizerV2 import SimpleTokenizerV2

tokenizer = SimpleTokenizerV2(vocab)
text1 = "Hello, Is this the right way to do it?"
text2 = "In the sunlit terraces of the palace."
text = "<|endoftext|>".join((text1, text2))
print(tokenizer.encode(text))
print(tokenizer.decode(tokenizer.encode(text)))
```
