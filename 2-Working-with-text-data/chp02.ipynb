## 2.1 Reading a short story

```python
# Reading text from file and displaying it onto the screen 
with open("the-verdict.txt", 'r', encoding="utf-8") as f:
    raw_text = f.read()

print(f"Total number of characters: {len(raw_text)}")
print(f"{raw_text[:99]}")
```

```python
# Splitting text manually using re built-in python library
import re
preprocessed = re.split(r'([,.:;?_!"()\']|--|\s)', raw_text) 
print(preprocessed[:99])
```

```python
# removing whitespaces to cut computing requirements
preprocessed = [item.strip() for item in preprocessed if item.strip()]
print(len(preprocessed))
print(preprocessed[:99])
```

```python
# sorting tokens in alphabetical order
all_words = sorted(set(preprocessed))
vocab_size = len(all_words)
print(vocab_size)
print(all_words[:30])
```

```python
# Creating vocab
vocab = {token:integer for integer,token in enumerate(all_words)}
for i, item in enumerate(vocab.items()):
    print(item)
    if i>=50:
        break
```
